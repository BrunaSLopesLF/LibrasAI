```python
# 2. Configurar o ngrok (Proxy Reverso) e subir o FastAPI localmente
import nest_asyncio
from pyngrok import ngrok, conf
import uvicorn
import sys

# ⚠️ COLOQUE SEU TOKEN DO NGROK AQUI (Crie conta em ngrok.com)
NGROK_TOKEN = "SEU_TOKEN_AQUI"
if NGROK_TOKEN != "SEU_TOKEN_AQUI":
    conf.get_default().auth_token = NGROK_TOKEN

port = 8000
public_url = ngrok.connect(port)
print("=" * 60)
print(f"🔥 BACKEND LIBRAS AI EXPOSTO PUBLICAMENTE EM:\n   {public_url}")
print("=" * 60)
print("No seu arquivo app.js, copie essa URL (trocando o http/https por ws/wss) no WebSocket!")

# Aplicar patch para rodar Uvicorn de forma assíncrona no Colab (que já tem event loop rodando)
nest_asyncio.apply()

print("Iniciando API no Google Colab...")
if __name__ == '__main__':
    # Aqui chamamos o Uvicorn indicando 'server:app'
    uvicorn.run("server:app", host="127.0.0.1", port=port)
```
